```python
Q1 = df_ham[['Sales', 'Profit']].quantile(0.25)
Q3 = df_ham[['Sales', 'Profit']].quantile(0.75)

IQR = Q3 - Q1

alt_sinir = Q1 - 1.5 * IQR
ust_sinir = Q3 + 1.5 * IQR

print("Alt sınırlar:")
print(alt_sinir)

print("\nÜst sınırlar:")
print(ust_sinir)
```

```python
aykiri_sales = df_ham[
    (df_ham['Sales'] < alt_sinir['Sales']) |
    (df_ham['Sales'] > ust_sinir['Sales'])
]

aykiri_profit = df_ham[
    (df_ham['Profit'] < alt_sinir['Profit']) |
    (df_ham['Profit'] > ust_sinir['Profit'])
]

print("Sales aykırı değer sayısı:", len(aykiri_sales))
print("Profit aykırı değer sayısı:", len(aykiri_profit))
```

```python
aykiri_sales[['Order ID', 'Product Name', 'Category', 'Sales', 'Quantity', 'Discount', 'Profit']].sort_values(
    'Sales', ascending=False
).head(10)
```

## Kritere Göre Tekrar Eden Kayıt Kontrolü

```python
tekrar_sayisi = df_ham.duplicated(subset=['Customer ID', 'Order Date', 'Product ID']).sum()
print(f"Müşteri + sipariş tarihi + ürün bazında tekrar eden kayıt sayısı: {tekrar_sayisi}")
```

##  Kategori Tutarlılığı Kontrolü




```python
kategorik_sutunlar = ['Category', 'Sub-Category', 'Region', 'Segment', 'Ship Mode']

for sutun in kategorik_sutunlar:
    print(f"{sutun}: {sorted(df_ham[sutun].unique())}")
    print()
```

```python
# Baştaki/sondaki boşluk kontrolü (tüm metin sütunları için)
metin_sutunlari = df_ham.select_dtypes(include='object').columns
bosluk_bulundu = False

for sutun in metin_sutunlari:
    bosluk_sorunu = (df_ham[sutun] != df_ham[sutun].str.strip()).sum()
    if bosluk_sorunu > 0:
        bosluk_bulundu = True
        print(f"{sutun}: {bosluk_sorunu} satırda baştaki/sondaki boşluk sorunu var")

if not bosluk_bulundu:
    print("Boşluk sorunu bulunmadı.")
```

```python
# Product Name sütunundaki boşluk sorununu düzeltelim
df_ham['Product Name'] = df_ham['Product Name'].str.strip()

kalan_sorun = (df_ham['Product Name'] != df_ham['Product Name'].str.strip()).sum()
print(f"Düzeltme sonrası 'Product Name' sütununda kalan boşluk sorunu: {kalan_sorun}")
```

##  Aykırı Değerlerin Ele Alınması,


```python
df_ham['Sales_Aykiri'] = (df_ham['Sales'] < alt_sinir['Sales']) | (df_ham['Sales'] > ust_sinir['Sales'])
df_ham['Profit_Aykiri'] = (df_ham['Profit'] < alt_sinir['Profit']) | (df_ham['Profit'] > ust_sinir['Profit'])

print("Sales aykırı olarak işaretlenen satır sayısı:", df_ham['Sales_Aykiri'].sum())
print("Profit aykırı olarak işaretlenen satır sayısı:", df_ham['Profit_Aykiri'].sum())
```

##  İndeks Düzenleme

```python
df_ham = df_ham.reset_index(drop=True)
print("İndeks sıfırlandı. İndeks aralığı:", df_ham.index.min(), "-", df_ham.index.max())
```

##  Temizlenmiş Veriyi Kaydetme



```python
df_ham.to_csv('Superstore_temiz.csv', index=False)
print("Temizlenmiş veri 'Superstore_temiz.csv' olarak kaydedildi.")
print("\nSon durumun genel bilgisi:")
df_ham.info()
```

##  Uçtan Uca Temizleme Fonksiyonu


```python
def veriyi_temizle(df_ham_girdi):

    df = df_ham_girdi.copy()


    df['Order Date'] = pd.to_datetime(df['Order Date'])
    df['Ship Date'] = pd.to_datetime(df['Ship Date'])


    df['Order Year'] = df['Order Date'].dt.year
    df['Order Month'] = df['Order Date'].dt.month
    df['Shipping Duration'] = df['Ship Date'] - df['Order Date']

    df['Product Name'] = df['Product Name'].str.strip()

    Q1 = df[['Sales', 'Profit']].quantile(0.25)
    Q3 = df[['Sales', 'Profit']].quantile(0.75)
    IQR = Q3 - Q1
    alt = Q1 - 1.5 * IQR
    ust = Q3 + 1.5 * IQR
    df['Sales_Aykiri'] = (df['Sales'] < alt['Sales']) | (df['Sales'] > ust['Sales'])
    df['Profit_Aykiri'] = (df['Profit'] < alt['Profit']) | (df['Profit'] > ust['Profit'])

    df = df.reset_index(drop=True)

    return df

df_test = veriyi_temizle(pd.read_csv('Superstore.csv', encoding='latin1'))
print("Fonksiyon testi - boyut:", df_test.shape)
df_test.head()
```

## Özet

Bu çalışmada gerçek Superstore satış verisi üzerinde pandas ile veri temizleme ve analiz sürecini uçtan uca uyguladık:

| Adım | Kullanılan Yöntem |
|---|---|
| Genel veri sağlığı kontrolü | `.info()`, `.isnull().sum()`, `.duplicated()` |
| Tarih sütunlarının tipi düzeltme | `pd.to_datetime()` |
| Tarihten yeni sütun türetme | `.dt.year`, `.dt.month`, tarih farkı |
| Aykırı değer tespiti | IQR yöntemi (`quantile`, `IQR`) |
| Aykırı değerleri bayraklama | Boole sütun ekleme |
| Kategori tutarlılığı kontrolü | `.unique()`, `.str.strip()` (Product Name'de 16 satırda boşluk bulundu ve düzeltildi) |
| İndeks düzenleme | `.reset_index(drop=True)` |
| Temiz veriyi kaydetme | `.to_csv()` |
| Uçtan uca fonksiyon | Tüm adımları tek fonksiyonda toplama |
